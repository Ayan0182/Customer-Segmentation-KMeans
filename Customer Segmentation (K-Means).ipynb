# Customer Segmentation using K-Means

```python
import pandas as pd
import numpy as np
import warnings 
warnings.filterwarnings("ignore")
```

# Data Loading

```python
df = pd.read_csv("Mall_Customers.csv")
```

```python
df.head()
```

```python
print(df.columns)
print(df.describe())
print(df.isnull().sum())
```

# Data Visualization

```python
import matplotlib.pyplot as plt

plt.scatter(df['Annual Income (k$)'], df['Spending Score (1-100)'])
plt.xlabel('Annual Income')
plt.ylabel('Spending Score')
plt.title('Customer Data Visualization')
plt.show()
```

# Clustering

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

x = df[['Annual Income (k$)', 'Spending Score (1-100)']]

kmeans = KMeans(n_clusters=5)
kmeans.fit(x)

y = kmeans.labels_

plt.scatter(x['Annual Income (k$)'], x['Spending Score (1-100)'], c=y)

plt.xlabel("Income")
plt.ylabel("Spending Score")
plt.title("Customer Groups")

plt.show()
```

```python
# Insightss

Group 1: High income, high spending - target customers
Group 2: High income, low spending - marketing needed
Group 3: Low income, high spending - impulsive buyers
Group 4: Low income, low spending - low priority
Group 5: Average customers - maintain engagement
```
